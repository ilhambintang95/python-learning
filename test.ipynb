```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

t = np.arange(0, 60)                       # bulan
q = 1000 * np.exp(-0.05 * t)               # rate, Arps exponential (contoh)
df = pd.DataFrame({"bulan": t, "rate": q})

df.plot(x="bulan", y="rate", logy=True, title="Tes decline eksponensial")
plt.show()
```
